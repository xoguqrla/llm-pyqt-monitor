```python
import pyvista as pv

plotter = pv.Plotter()
plotter.show()  # 반드시 창을 띄워야 context가 생성됨

ren_win = plotter.ren_win
try:
    # VTK의 GetOpenGLVersion()은 인자를 요구하므로, 아래처럼 빈 문자열을 넘깁니다.
    version_str = ""
    ren_win.GetOpenGLVersion(version_str, version_str)
    print("OpenGL Version:", version_str)
except Exception as e:
    print("OpenGL 버전 정보를 가져올 수 없습니다:", e)
```
