```python

# 데이터 불러오기
df = pd.read_csv('001_1_data.csv')

# load와 mpa 컬럼 추출
load = df['LOAD']
mpa = df['MPA']

# 피어슨 상관계수 계산
corr, p_value = pearsonr(load, mpa)

print(f'피어슨 상관계수: {corr:.4f}')
print(f'p-value: {p_value:.4e}')
```
